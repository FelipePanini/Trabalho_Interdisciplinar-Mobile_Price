```python
# ==========================================================
# GRÁFICO COMPARATIVO — ACURÁCIA POR VALOR DE K
# ==========================================================

import matplotlib.pyplot as plt

# Resultados do KNN Manual
k_manual   = [1, 5, 10, 15, 20, 30, 50, 60, 65, 70, 250]
acc_manual = [72.00, 78.25, 79.00, 81.00, 81.25, 81.50, 82.00, 82.25, 81.50, 80.50, 76.25]

# Resultados do KNN com scikit-learn
k_biblio   = [1, 5, 15, 25, 50, 100, 150]
acc_biblio = [71.00, 76.00, 78.00, 79.00, 80.00, 76.75, 76.00]

plt.figure(figsize=(10, 5))

plt.plot(k_manual, acc_manual,
         marker='o', color='steelblue',
         label='KNN Manual', linewidth=2)

plt.plot(k_biblio, acc_biblio,
         marker='s', color='darkorange',
         label='KNN Scikit-learn', linewidth=2, linestyle='--')

# Marca os melhores pontos
plt.scatter([60], [82.25], color='steelblue', s=120, zorder=5)
plt.annotate('K=60\n82,25%', xy=(60, 82.25),
             xytext=(75, 82.8), fontsize=9, color='steelblue')

plt.scatter([50], [80.00], color='darkorange', s=120, zorder=5)
plt.annotate('K=50\n80,00%', xy=(50, 80.00),
             xytext=(60, 79.0), fontsize=9, color='darkorange')

plt.title('Comparação de Acurácia — KNN Manual vs. Scikit-learn')
plt.xlabel('Valor de K')
plt.ylabel('Acurácia (%)')
plt.ylim(65, 88)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

## 3. Análise das diferenças observadas

### Acurácia

O KNN Manual obteve acurácia ligeiramente superior (**82,25%** vs **80,00%**). Isso ocorre por uma diferença na forma como os dados foram divididos:

- O **KNN Manual** usa divisão sequencial — os primeiros 80% dos dados (após embaralhamento com `random_state=42`) formam o treino.
- O **scikit-learn** usa divisão estratificada (`stratify=y`), garantindo proporção igual de cada classe em treino e teste.

A divisão estratificada é tecnicamente mais correta para classificação multiclasse, mesmo que o resultado de acurácia seja um pouco menor neste caso específico.

### Facilidade de implementação

A implementação manual exigiu escrever do zero a lógica de distância euclidiana, ordenação, seleção de vizinhos e votação por maioria — cerca de 80 linhas de código com comentários. Já com o scikit-learn, o mesmo resultado foi obtido em aproximadamente 15 linhas, aproveitando funções prontas e otimizadas.

### Aprendizado x Praticidade

Apesar de mais trabalhosa, a implementação manual é fundamental para compreender **como o algoritmo realmente funciona** — cada etapa do cálculo fica explícita no código. A biblioteca, por outro lado, é a escolha ideal para projetos reais, por oferecer mais recursos (matriz de confusão, relatório de classificação) com menos esforço.

### Conclusão

Ambas as abordagens confirmaram que, para este dataset e com os atributos selecionados (`battery_power`, `clock_speed`, `n_cores`, `ram`), o KNN consegue classificar corretamente a faixa de preço de celulares com **acurácia entre 80% e 82%**. O atributo com maior impacto no resultado foi a **RAM**, que apresenta alta correlação com a variável alvo `price_range`.
