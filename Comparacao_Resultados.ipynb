## 1. Exploração de Atributos

Antes de definir a configuração final, ambas as implementações foram testadas com três grupos distintos de atributos, com o objetivo de identificar quais variáveis contribuem mais para a classificação correta da faixa de preço.

## 2. Resultados por valor de K — Configuração final (8 colunas)

Ambas as implementações foram testadas com diferentes valores de K, utilizando os 8 atributos selecionados e a mesma proporção de treino/teste (80%/20%).

### KNN Manual

| K | Acurácia | F1-score |
|---|----------|----------|
| 1 | 61,25% | 0,61 |
| 9 | 69,50% | 0,70 |
| **15** | **73,50%** | **0,74** |
| 21 | 72,50% | 0,72 |
| 25 | 72,00% | 0,72 |
| 31 | 71,00% | 0,71 |

✅ **Melhor resultado manual: K = 15 → Acurácia de 73,50% / F1-score 0,74**

---

### KNN com scikit-learn

| K | Acurácia | F1-score |
|---|----------|----------|
| 1 | 64,00% | 0,64 |
| 9 | 67,25% | 0,68 |
| 15 | 71,25% | 0,72 |
| **19** | **73,75%** | **0,74** |
| 25 | 72,25% | 0,72 |

✅ **Melhor resultado com biblioteca: K = 19 → Acurácia de 73,75% / F1-score 0,74**

````python
# ==========================================================
# GRÁFICO COMPARATIVO — ACURÁCIA POR VALOR DE K
# ==========================================================

import matplotlib.pyplot as plt

# Resultados do KNN Manual (8 colunas)
k_manual   = [1, 9, 15, 21, 25, 31]
acc_manual = [61.25, 69.50, 73.50, 72.50, 72.00, 71.00]

# Resultados do KNN com scikit-learn (8 colunas)
k_biblio   = [1, 9, 15, 19, 25]
acc_biblio = [64.00, 67.25, 71.25, 73.75, 72.25]

plt.figure(figsize=(10, 5))

plt.plot(k_manual, acc_manual,
         marker='o', color='steelblue',
         label='KNN Manual', linewidth=2)

plt.plot(k_biblio, acc_biblio,
         marker='s', color='darkorange',
         label='KNN Scikit-learn', linewidth=2, linestyle='--')

# Marca os melhores pontos
plt.scatter([15], [73.50], color='steelblue', s=150, zorder=5)
plt.annotate('K=15\n73,50%', xy=(15, 73.50),
             xytext=(16.5, 74.5), fontsize=9, color='steelblue')

plt.scatter([19], [73.75], color='darkorange', s=150, zorder=5)
plt.annotate('K=19\n73,75%', xy=(19, 73.75),
             xytext=(20.5, 72.2), fontsize=9, color='darkorange')

plt.title('Comparação de Acurácia — KNN Manual vs. Scikit-learn (8 atributos)')
plt.xlabel('Valor de K')
plt.ylabel('Acurácia (%)')
plt.ylim(55, 80)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
````

## 3. Comparação direta — Melhor resultado de cada abordagem

| Critério | KNN Manual (K=15) | KNN Scikit-learn (K=19) |
|----------|-------------------|-------------------------|
| Acurácia | 73,50% | **73,75%** |
| F1-score | 0,74 | 0,74 |
| Valor de K ideal | 15 | 19 |
| Atributos utilizados | 8 colunas | 8 colunas |
| Métricas disponíveis | Acurácia, F1, matriz de confusão | Acurácia, precisão, recall, F1, matriz de confusão |
| Linhas de código | ~80 linhas | ~15 linhas |
| Facilidade de uso | Média — requer implementar tudo do zero | Alta — chamada de função única |
| Transparência | Alta — cada passo é visível no código | Baixa — lógica interna abstraída |
| Tempo de execução | Mais lento (loop Python puro) | Mais rápido (otimizado em C++) |

````python
# ==========================================================
# GRÁFICO — EVOLUÇÃO DA ACURÁCIA POR CONFIGURAÇÃO DE ATRIBUTOS
# ==========================================================

import matplotlib.pyplot as plt
import numpy as np

configuracoes = ['Todas\n(20 colunas)', '14 colunas', '8 colunas']

melhor_manual  = [57.25, 63.75, 73.50]
melhor_biblio  = [49.25, 57.25, 73.75]

x = np.arange(len(configuracoes))
largura = 0.35

fig, ax = plt.subplots(figsize=(9, 5))

barras_manual = ax.bar(x - largura/2, melhor_manual, largura,
                       label='KNN Manual', color='steelblue', alpha=0.85)
barras_biblio = ax.bar(x + largura/2, melhor_biblio, largura,
                       label='KNN Scikit-learn', color='darkorange', alpha=0.85)

# Adiciona os valores nas barras
for barra in barras_manual:
    ax.text(barra.get_x() + barra.get_width()/2, barra.get_height() + 0.5,
            f'{barra.get_height():.2f}%', ha='center', va='bottom', fontsize=9)

for barra in barras_biblio:
    ax.text(barra.get_x() + barra.get_width()/2, barra.get_height() + 0.5,
            f'{barra.get_height():.2f}%', ha='center', va='bottom', fontsize=9)

ax.set_title('Evolução da Acurácia por Configuração de Atributos')
ax.set_ylabel('Melhor Acurácia (%)')
ax.set_xticks(x)
ax.set_xticklabels(configuracoes)
ax.set_ylim(40, 82)
ax.legend()
ax.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
````

## 4. Análise das diferenças observadas

### Acurácia

As duas implementações atingiram resultados praticamente idênticos — **73,50% (manual)** vs **73,75% (biblioteca)** — com o mesmo conjunto de 8 atributos. A diferença de apenas 0,25 ponto percentual confirma que a implementação manual está matematicamente correta e equivalente à solução da biblioteca.

Vale destacar que o K ideal diferiu entre as abordagens (K=15 no manual e K=19 na biblioteca). Isso ocorre porque a divisão dos dados é feita de forma diferente:

- O **KNN Manual** usa divisão sequencial — os primeiros 80% dos dados (após embaralhamento com `random_state=42`) formam o treino.
- O **scikit-learn** usa divisão **estratificada** (`stratify=y`), garantindo proporção igual de cada classe em treino e teste, o que torna o conjunto de treino ligeiramente diferente.

### Seleção de atributos

Um dos principais aprendizados do trabalho foi que **mais atributos não significa melhor desempenho**. O modelo com todas as 20 colunas apresentou acurácia inferior (~57%) em comparação ao modelo com apenas 8 colunas cuidadosamente selecionadas (~73%). Atributos com baixa correlação com a variável alvo introduzem ruído e prejudicam o algoritmo KNN, que é sensível à distância entre os pontos.

### Facilidade de implementação

A implementação manual exigiu escrever do zero a lógica de distância euclidiana, ordenação, seleção de vizinhos e votação por maioria — cerca de 80 linhas de código com comentários. Já com o scikit-learn, o mesmo resultado foi obtido em aproximadamente 15 linhas, aproveitando funções prontas e otimizadas.

### Aprendizado x Praticidade

Apesar de mais trabalhosa, a implementação manual é fundamental para compreender **como o algoritmo realmente funciona** — cada etapa do cálculo fica explícita no código. A biblioteca, por outro lado, é a escolha ideal para projetos reais, por oferecer mais recursos (matriz de confusão, relatório de classificação completo) com menos esforço.

### Conclusão

Ambas as abordagens confirmaram que, para este dataset e com os 8 atributos selecionados (`battery_power`, `clock_speed`, `storage`, `n_cores`, `primary_camera`, `px_width`, `ram`, `talk_time`), o KNN consegue classificar corretamente a faixa de preço de celulares com **acurácia de aproximadamente 73-74%**. A convergência dos resultados entre as duas implementações valida a corretude do código desenvolvido manualmente.
